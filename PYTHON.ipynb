```python
contact_list=[
    {"NAME":"JAMIE","HOUSE":"LANNISTER","TITLE":"THE KINGSLAYER"},
    {"NAME":"AARYA","HOUSE":"STARK","TITLE":"NO ONE"},
    {"NAME":"DANERYAS","HOUSE":"TARGARYEN","TITLE":"MOTHER OF DRAGONS"}
]

def display_contact_list(input_list):
    if not input_list:
        print("No contacts to display.")
        return
    for contact in input_list:
      print("=====INFO=====")
      print()
      print("Name:",contact["NAME"])
      print("House:",contact["HOUSE"])
      print("Title:",contact["TITLE"])
      print()
      print("=====END=====")

def add_contact(input_list):
        new_contact_name=input("Name of the contact: ")
        new_contact_house=input("House of the contact: ")
        new_contact_title=input("Title of the contact: ")

        new_contact={
            "NAME":new_contact_name,
            "HOUSE":new_contact_house,
            "TITLE":new_contact_title
        }
        input_list.append(new_contact)
        print("Contact added successfully!")
        return input_list

def search_contacts(input_list):
    search_term = input("Enter name, house, or title to search: ").upper()
    found_contacts = []
    for contact in input_list:
        if search_term in contact["NAME"].upper() or \
           search_term in contact["HOUSE"].upper() or \
           search_term in contact["TITLE"].upper():
            found_contacts.append(contact)
    if found_contacts:
        print("\n--- Search Results ---")
        display_contact_list(found_contacts)
    else:
        print("No contacts found matching your search.")

def new_expanded_menu():
  print("\n"+"="*40)
  print("Character Book Menu")
  print("="* 40)
  print("1. View all Contacts")
  print("2. Add new Contacts")
  print("3. Search Contacts")
  print("4. Exit")
  print("="* 40)

  choice=input("\nWhat would you like to do? ")
  return choice

print("Welcome to our Fancy Book")
keep_going=True
while keep_going:
    user_input=new_expanded_menu()
    print("You want to do: ",user_input)
    if user_input == '1':
        display_contact_list(contact_list)
    elif user_input == '2':
        contact_list = add_contact(contact_list)
    elif user_input == '3':
        search_contacts(contact_list)
    elif user_input == '4':
        keep_going = False
        print("Exiting program...")
    else:
        print("Invalid choice. Please try again.")
```
